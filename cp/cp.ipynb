```python
def merge_sorted_arrays(nums1, nums2):
    n1 = len(nums1)
    n2 = len(nums2)
    merged = [0] * (n1 + n2)

    i = j = k = 0

    # Trộn hai mảng theo thứ tự không giảm
    while i < n1 and j < n2:
        if nums1[i] <= nums2[j]:
            merged[k] = nums1[i]
            i += 1
        else:
            merged[k] = nums2[j]
            j += 1
        k += 1

    # nums1
    while i < n1:
        merged[k] = nums1[i]
        i += 1
        k += 1
        
    # nums2
    while j < n2:
        merged[k] = nums2[j]
        j += 1
        k += 1

    return merged
```

```python
nums1 = [1, 2, 3]
nums2 = [2, 5, 6]
result = merge_sorted_arrays(nums1, nums2)
print(result)
```
